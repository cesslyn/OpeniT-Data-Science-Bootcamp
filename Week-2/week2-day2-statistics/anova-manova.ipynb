```python
from pathlib import Path

import pandas as pd
import plotly.express as px
from scipy.stats import f_oneway
from statsmodels.multivariate.manova import MANOVA

ACTIVITY_DATA_DIR = Path("Activity Data")

DATA_FILE = ACTIVITY_DATA_DIR / "anova-sample-data.csv"
if not DATA_FILE.exists():
    raise FileNotFoundError(f"Missing data file: {DATA_FILE}")
```

# ANOVA: campaign channel and conversion rate

```python
anova_df = pd.read_csv(ACTIVITY_DATA_DIR / "anova-sample-data.csv")
anova_df.head()
```

```python
anova_df.groupby("channel")["conversion_rate_pct"].agg(["count", "mean", "std"]).round(2)
```

```python
groups = [group["conversion_rate_pct"].to_numpy() for _, group in anova_df.groupby("channel")]
anova_result = f_oneway(*groups)

print(f"F-statistic: {anova_result.statistic:.4f}")
print(f"p-value: {anova_result.pvalue:.6f}")
```

```python
fig = px.box(
    anova_df,
    x="channel",
    y="conversion_rate_pct",
    color="channel",
    points="all",
    title="ANOVA: Conversion Rate by Campaign Channel",
)

fig.update_layout(template="plotly_white", showlegend=False)
fig.show()
```
