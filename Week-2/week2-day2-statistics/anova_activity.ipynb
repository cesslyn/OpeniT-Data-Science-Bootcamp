Complete the missing code to test whether conversion rates differ across campaign channels.


## Setup


```python
from pathlib import Path

import pandas as pd
import plotly.express as px
from scipy.stats import f_oneway


ACTIVITY_DATA_DIR = Path("Activity Data")

DATA_FILE = ACTIVITY_DATA_DIR / "anova-sample-data.csv"
if not DATA_FILE.exists():
    raise FileNotFoundError(f"Missing data file: {DATA_FILE}")
```

## 1. Load And Inspect The Data

Load the local ANOVA sample data and inspect the campaign channel and conversion-rate columns.


```python
# Load the ANOVA dataset from DATA_FILE.
# Use pd.read_csv and pass DATA_FILE inside the parentheses.
anova_df = pd.read_csv(ACTIVITY_DATA_DIR / "anova-sample-data.csv")

# Display the first few rows so you can check the column names.
anova_df.head()
```

```python
# Summarize conversion rates by campaign channel.
# Step 1: group by the channel column.
# Step 2: select the conversion_rate_pct column.
# Step 3: calculate count, mean, and standard deviation.
channel_summary = anova_df.groupby("channel")["conversion_rate_pct"].agg(["count", "mean", "std"]).round(2)
channel_summary
```

## 2. Run One-Way ANOVA

Create one conversion-rate array per channel, then pass those arrays into `f_oneway`.


```python
# Create one NumPy array of conversion rates for each channel.
# Start with an empty list, then append one group at a time.
groups = []

# Group the data by channel.
for channel, group in anova_df.groupby("channel"):
    # Select conversion_rate_pct from this channel's rows.
    conversion_rates = group["conversion_rate_pct"].to_numpy()

    # Add this channel's conversion-rate array to the groups list.
    groups.append(conversion_rates)

# Run one-way ANOVA.
# Use *groups to pass each channel array into f_oneway as a separate argument.
anova_result = f_oneway(*groups)

print(f"F-statistic: {anova_result.statistic:.4f}")
print(f"p-value: {anova_result.pvalue:.15f}")
```

## 3. Visualize The Group Differences

Use a box plot to compare the conversion-rate distribution for each campaign channel.


```python
# Create a box plot comparing conversion_rate_pct across channels.
fig = px.box(
    anova_df,
    x="channel",
    y="conversion_rate_pct",
    color="channel",
    points="all",
    title="ANOVA: Conversion Rate by Campaign Channel",
)

fig.update_layout(template="plotly_white", showlegend=False)
fig.show()
```
